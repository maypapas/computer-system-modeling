```python

import pandas as pd
from google.colab import files

# Upload the 4 CSV files
files.upload()

# Read CSV files
def read_data(filename):
    df = pd.read_csv(
        filename,
        sep=None,
        engine="python"
    )

    df.columns = df.columns.str.strip()
    df = df.loc[:, ~df.columns.str.startswith("Unnamed")]
    df = df.dropna(axis=1, how="all")

    if "classname" not in df.columns:
        raise ValueError("Missing classname: " + filename)

    df["classname"] = df["classname"].astype(str).str.strip()

    if df["classname"].duplicated().any():
        raise ValueError("Duplicate classname: " + filename)

    return df


ck = read_data("single-version-ck-oo.csv")
bug = read_data("bug-metrics.csv")
change = read_data("change-metrics.csv")
ent = read_data("ent.csv")

# Post-release defect labels must not be input features
target_columns = {
    "bugs",
    "nontrivialbugs",
    "majorbugs",
    "criticalbugs",
    "highprioritybugs"
}

def get_features(df, prefix):
    keep = [
        c for c in df.columns
        if c == "classname"
        or c.lower() not in target_columns
    ]

    df = df[keep].copy()

    df = df.rename(columns={
        c: prefix + "_" + c
        for c in df.columns
        if c != "classname"
    })

    return df


# Create defect classification label
if "bugs" not in change.columns:
    raise ValueError("Missing bugs column in change-metrics.csv")

counts = pd.to_numeric(change["bugs"], errors="raise")

if counts.isna().any():
    raise ValueError("Missing post-release defect counts")

JDT = change[["classname"]].copy()
JDT["class"] = (counts > 0).astype(int)

# Combine metric groups
datasets = [
    (get_features(ck, "ck_oo"), "CK/OO"),
    (get_features(bug, "previous_bug"), "Previous Bug"),
    (get_features(ent, "entropy"), "Entropy")
]

# Merge using classname
for data, name in datasets:

    matches = JDT["classname"].isin(data["classname"]).sum()
    print(name, "- Matched classes:", matches)

    JDT = JDT.merge(
        data,
        on="classname",
        how="left",
        validate="one_to_one"
    )

# Convert features into numeric values
for col in JDT.columns:
    if col not in ["classname", "class"]:
        JDT[col] = pd.to_numeric(
            JDT[col], errors="coerce"
        )

# Remove entirely empty columns
JDT = JDT.dropna(axis=1, how="all")

# Put class at the end
label = JDT.pop("class")
JDT["class"] = label

# Save JDT.csv
JDT.to_csv("JDT.csv", index=False)

print("\nJDT.csv created successfully")
print("Rows:", JDT.shape[0])
print("Features:", JDT.shape[1] - 2)
print("Defective classes:", JDT["class"].sum())

print(JDT.head())

files.download("JDT.csv")
```
