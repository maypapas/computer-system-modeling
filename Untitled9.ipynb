```python

%pip install scikit-optimize
# Stage 1: Hyperparameter Tuning
# Stage 2: XGBoost Training and Evaluation

!pip install -q xgboost scikit-learn

import numpy as np
import pandas as pd
import os
import zipfile
import matplotlib.pyplot as plt

from xgboost import XGBRegressor
from sklearn.model_selection import GridSearchCV, KFold
from sklearn.metrics import make_scorer
from scipy.stats import t
from google.colab import files
from IPython.display import display

from skopt import BayesSearchCV
from skopt.space import Integer, Real



# 1. UPLOAD DATASETS

files.upload()

systems = {
    "Apache": ("Apache", [9, 18, 27, 36, 45]),
    "x264": ("x264", [16, 32, 48, 64, 80]),
    "LLVM": ("LLVM", [11, 22, 33, 44, 55]),
    "BDB-C": ("BDBC", [18, 36, 54, 72, 90]),
    "BDB-J": ("BDBJ", [26, 52, 78, 104, 130]),
    "SQLite": ("SQL", [39, 78, 117, 156, 195]),
    "DUNE MGS": ("Dune", [49, 78, 240, 375]),
    "HIPAcc": ("hipacc", [261, 528, 736, 1281]),
    "HSMGP": ("hsmgp", [77, 173, 384, 480]),
    "JavaGC": ("javagc", [423, 534, 855, 2571]),
    "SaC": ("sac", [2060, 2295, 2499, 3261])
}

datasets = {}

for system, (filename, sizes) in systems.items():

    df = pd.read_csv(filename + "_AllNumeric.csv")

    X = df.iloc[:, :-1].values
    y = df.iloc[:, -1].values

    if not np.isfinite(X).all() or not np.isfinite(y).all():
        raise ValueError(f"Invalid values in {system}")

    if np.any(y == 0):
        raise ValueError(f"Zero target values in {system}")

    datasets[system] = (X, y)

    print(f"{system}: {len(y)} configurations")

print("\nAll datasets loaded successfully.")

```

```python
# 2. DEFINE MRE AND RANDOM SPLITTING

def mre(y_true, y_pred):

    return np.mean(
        np.abs(y_true - y_pred) / np.abs(y_true)
    ) * 100


def get_split(system, size, repeat):

    X, y = datasets[system]

    seed = (
        2026
        + 10000 * list(systems).index(system)
        + 100 * size
        + repeat
    )

    rng = np.random.default_rng(seed)

    indices = rng.permutation(len(y))

    train_idx = indices[:size]
    test_idx = indices[size:]

    # Limit testing configurations for large datasets
    if system in ["JavaGC", "SaC"]:
        test_idx = test_idx[:15000]

    return train_idx, test_idx
```

```python

from skopt import BayesSearchCV
from skopt.space import Integer, Real
from sklearn.model_selection import KFold
from sklearn.metrics import make_scorer
from xgboost import XGBRegressor

# STAGE 1: BAYESIAN OPTIMIZATION

print("\n========== STAGE 1: BAYESIAN OPTIMIZATION ==========")

# Hyperparameter search space
search_space = {
    "n_estimators": Integer(100, 300),
    "max_depth": Integer(2, 6),
    "learning_rate": Real(0.01, 0.2, prior="log-uniform")
}

# Lower MRE is better
scorer = make_scorer(
    mre,
    greater_is_better=False
)

all_tuning = []
best_results = []
best_params = {}

for system, (_, sample_sizes) in systems.items():

    X, y = datasets[system]

    for size in sample_sizes:

        # Tune on the first training sample
        train_idx, _ = get_split(system, size, 1)

        X_train = X[train_idx]
        y_train = y[train_idx]

        model = XGBRegressor(
            objective="reg:squarederror",
            tree_method="hist",
            n_jobs=2,
            random_state=42
        )

        # Bayesian Optimization
        bayes = BayesSearchCV(
            estimator=model,
            search_spaces=search_space,
            n_iter=20,
            scoring=scorer,
            cv=KFold(
                n_splits=3,
                shuffle=True,
                random_state=42
            ),
            n_jobs=1,
            random_state=42,
            refit=True
        )

        # Perform Bayesian Optimization
        bayes.fit(X_train, y_train)

        # Save all tuning results
        for params, score in zip(
            bayes.cv_results_["params"],
            bayes.cv_results_["mean_test_score"]
        ):

            all_tuning.append({
                "System": system,
                "Sample Size": size,
                **params,
                "Validation MRE (%)": -score
            })

        # Save best hyperparameters
        best_params[(system, size)] = bayes.best_params_

        best_results.append({
            "System": system,
            "Sample Size": size,
            **bayes.best_params_,
            "Best Validation MRE (%)": -bayes.best_score_
        })

        print(
            f"{system} | {size} samples | "
            f"Best MRE: {-bayes.best_score_:.2f}% | "
            f"Best Params: {bayes.best_params_}"
        )


```

```python
# 3. DISPLAY ALL TUNING RESULTS

tuning_df = pd.DataFrame(all_tuning)

print("\n========== ALL TUNING RESULTS ==========")

with pd.option_context(
    "display.max_rows", None,
    "display.max_columns", None
):
    display(tuning_df)

tuning_df.to_csv(
    "all_tuning_results.csv",
    index=False
)
```

```python
# 4. DISPLAY BEST HYPERPARAMETERS

best_df = pd.DataFrame(best_results)

print("\n========== BEST HYPERPARAMETERS ==========")

with pd.option_context("display.max_rows", None):
    display(best_df)

best_df.to_csv(
    "best_hyperparameters.csv",
    index=False
)
```

```python
# STAGE 2: FINAL XGBOOST TRAINING


# Run this section after reviewing tuning results.
print("\n========== STAGE 2: TRAINING ==========")

all_runs = []
final_results = []

for system, (_, sample_sizes) in systems.items():

    X, y = datasets[system]

    for size in sample_sizes:

        errors = []

        # Get best hyperparameters from Stage 1
        params = best_params[(system, size)]

        # Repeat experiment 30 times
        for repeat in range(1, 31):

            train_idx, test_idx = get_split(
                system,
                size,
                repeat
            )

            X_train = X[train_idx]
            y_train = y[train_idx]

            X_test = X[test_idx]
            y_test = y[test_idx]

            # Create XGBoost with best hyperparameters
            model = XGBRegressor(
                **params,
                objective="reg:squarederror",
                tree_method="hist",
                n_jobs=2,
                random_state=2026 + repeat
            )

            # Train model
            model.fit(X_train, y_train)

            # Predict testing dataset
            predictions = model.predict(X_test)

            # Calculate MRE
            test_mre = mre(y_test, predictions)

            errors.append(test_mre)

            # Save individual repetition result
            all_runs.append({
                "System": system,
                "Sample Size": size,
                "Repetition": repeat,
                "Test MRE (%)": test_mre
            })

        # Calculate average MRE
        mean_mre = np.mean(errors)

        # Calculate 95% confidence interval margin
        margin = (
            t.ppf(0.975, 29)
            * np.std(errors, ddof=1)
            / np.sqrt(30)
        )

        final_results.append({
            "System": system,
            "Sample Size": size,
            "Mean MRE (%)": mean_mre,
            "95% CI Margin": margin,
            "CI Lower": mean_mre - margin,
            "CI Upper": mean_mre + margin
        })

        print(
            f"{system:10s} | "
            f"Sample: {size:5d} | "
            f"MRE: {mean_mre:.3f}% "
            f"± {margin:.3f}%"
        )
```

```python
# 5. DISPLAY FINAL RESULTS


final_df = pd.DataFrame(final_results)
runs_df = pd.DataFrame(all_runs)

print("\n========== FINAL XGBOOST RESULTS ==========")

with pd.option_context("display.max_rows", None):
    display(final_df)

final_df.to_csv(
    "final_xgboost_results.csv",
    index=False
)

runs_df.to_csv(
    "all_1500_runs.csv",
    index=False
)
```

```python
# 6. SAVE AND DOWNLOAD ALL RESULTS

output_files = [
    "all_tuning_results.csv",
    "best_hyperparameters.csv",
    "final_xgboost_results.csv",
    "all_1500_runs.csv"
]

with zipfile.ZipFile(
    "XGBoost_DeepPerf_Results.zip",
    "w"
) as zipf:

    for filename in output_files:
        zipf.write(filename)

print("\nExperiment completed!")
print("Total tuning combinations:", len(tuning_df))
print("Total training experiments:", len(runs_df))
print("Total final settings:", len(final_df))

files.download("XGBoost_DeepPerf_Results.zip")
```
